```python
matrix = np.array([
    [10, 20, 30],
    [5,  5,  5]
])

print(matrix.sum(axis=0))
# [15, 25, 35] -> Мы "схлопнули" строки. Сумма по вертикали.

print(matrix.sum(axis=1))
```

```python
arr = np.arange(12) # [0, 1, ..., 11]

# Превращаем в матрицу 3x4
reshaped = arr.reshape(3, 4)

# Магия -1: "NumPy, посчитай сам"
# Я хочу 3 строки, а столбцов — сколько получится.
auto_reshaped = arr.reshape(3, -1)
# Результат тот же: (3, 4). Это очень часто используется в продакшене.

print(reshaped)
print(auto_reshaped)
```

```python
data = np.array([10, 20, -999, 30, -999, 40])

# 1. Создаем маску (True/False)
mask = (data != -999)

# 2. Чистим данные
clean_data = data[mask] # [10, 20, 30, 40]
print(clean_data)

# 3. Или заменяем ошибки на среднее (Data Imputation)
mean_val = np.mean(clean_data)
print(mean_val)
data[data == -999] = mean_val
print(data)
```

```python
heights = np.array([160, 170, 180])
mean = heights.mean() # 170.0
print(mean)

# Broadcasting в действии:
# NumPy вычитает скаляр 170.0 из КАЖДОГО элемента массива.
normalized = heights - mean
print(normalized)
# [-10, 0, 10]
```

```python
# Входы нейрона (inputs)
X = np.array([1, 2])
# Веса нейрона (weights)
W = np.array([0.5, 0.8])

# Формула нейрона: Y = X * W
# Поэлементно: (1*0.5) + (2*0.8) = 0.5 + 1.6 = 2.1
output = X @ W  # Одной операцией!

print(output) # 2.1
```

```python
# Входы нейрона (inputs)
X = np.array([1, 2])
# Веса нейрона (weights)
W = np.array([0.5, 0.8])

# Формула нейрона: Y = X * W
# Поэлементно: (1*0.5) + (2*0.8) = 0.5 + 1.6 = 2.1
X_and_W = X * W  # Одной операцией!

print(X_and_W) # 2.1
```

```python
# Эмулируем картинку 2x2 пикселя (RGB цвета)
# Shape: (2, 2, 3) -> 2 строки, 2 столбца, 3 канала цвета
image = np.array([
    [[255, 0, 0], [0, 255, 0]], # Красный, Зеленый
    [[0, 0, 255], [255, 255, 0]] # Синий, Желтый
])

# 1. Распакуем веса для каналов
weights = np.array([0.299, 0.587, 0.114])

# 2. Применяем Dot product вдоль последней оси (цвета)
# Мы умножаем RGB значения каждого пикселя на веса и складываем их.
grayscale_image = image @ weights

print("Было (RGB):\n", image.shape)      # (2, 2, 3)
print("Стало (Grayscale):\n", grayscale_image.shape) # (2, 2)
print("\nЗначения яркости:\n", grayscale_image)
```

```python
inventory = np.ones((4, 3)) # 4 склада, 3 товара
delivery_costs = np.array([10, 20, 30]) # Стоимость для каждого товара

# Работает! Вектор (3,) растянется до (4, 3)
total_cost = inventory + delivery_costs
print(total_cost)
```

```python
x = np.arange(5)
print(x)
y = np.empty(5) # Создаем пустую заготовку
print(y)

np.multiply(x, 10, out=y) # Результат сразу летит в y
print(y) # [ 0. 10. 20. 30. 40.]
```
