```python
data = np.array([10, 20, -999, 30, -999, 40])

# 1. Создаем маску (True/False)
mask = (data != -999)

# 2. Чистим данные
clean_data = data[mask] # [10, 20, 30, 40]
print(clean_data)

# 3. Или заменяем ошибки на среднее (Data Imputation)
mean_val = np.mean(clean_data)
print(mean_val)
data[data == -999] = mean_val
print(data)
```

```python
heights = np.array([160, 170, 180])
mean = heights.mean() # 170.0
print(mean)

# Broadcasting в действии:
# NumPy вычитает скаляр 170.0 из КАЖДОГО элемента массива.
normalized = heights - mean
print(normalized)
# [-10, 0, 10]
```

```python
# Входы нейрона (inputs)
X = np.array([1, 2])
# Веса нейрона (weights)
W = np.array([0.5, 0.8])

# Формула нейрона: Y = X * W
# Поэлементно: (1*0.5) + (2*0.8) = 0.5 + 1.6 = 2.1
output = X @ W  # Одной операцией!

print(output) # 2.1
```

```python
# Эмулируем картинку 2x2 пикселя (RGB цвета)
# Shape: (2, 2, 3) -> 2 строки, 2 столбца, 3 канала цвета
image = np.array([
    [[255, 0, 0], [0, 255, 0]], # Красный, Зеленый
    [[0, 0, 255], [255, 255, 0]] # Синий, Желтый
])

# 1. Распакуем веса для каналов
weights = np.array([0.299, 0.587, 0.114])

# 2. Применяем Dot product вдоль последней оси (цвета)
# Мы умножаем RGB значения каждого пикселя на веса и складываем их.
grayscale_image = image @ weights

print("Было (RGB):\n", image.shape)      # (2, 2, 3)
print("Стало (Grayscale):\n", grayscale_image.shape) # (2, 2)
print("\nЗначения яркости:\n", grayscale_image)
```
